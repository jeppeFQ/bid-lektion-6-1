## 7: Klassen passer på sig selv

Nu samler vi trin 3 og trin 4: reglen om de 50 ansatte skal håndhæves, **når objektet
bliver lavet**. Så kan ingen lave en `Arbejdsplads`, der bryder den.

`raise` stopper programmet med en fejl, man selv har formuleret.
`UgyldigArbejdsplads` er vores egen fejltype. Den "arver" fra `ValueError`, så den
opfører sig som en almindelig Python-fejl, men har sit eget navn.

> *Think Python kapitel 14 og 16 og `raise` fra kapitel 5.*

**Udfyld hullerne.**

```python
class UgyldigArbejdsplads(ValueError):
    """Tallene for en arbejdsplads giver ikke mening."""


class Arbejdsplads:
    """En arbejdsplads, som barometeret ser den."""

    def __init__(self, ansatte, antal, branche, sektor, region):
        if ansatte _______ 50:
            raise UgyldigArbejdsplads("Barometeret dækker kun arbejdspladser med mindst 50 ansatte.")

        for dimension in antal:
            if antal[dimension] _______ ansatte:
                raise UgyldigArbejdsplads(f"Der kan ikke være flere i gruppen {dimension} end der er ansatte.")

        self.ansatte = ansatte
        self.antal = antal
        self.branche = branche
        self.sektor = sektor
        self.region = region

    def andel(self, dimension):
        """Andelen i én dimension, som et tal mellem 0 og 1."""
        return self.antal[dimension] / self.ansatte

    def __str__(self):
        return f"{self.ansatte} ansatte i {self.branche}"


# Den her virker
skolen = Arbejdsplads(120, {"kvinder": 84, "seniorer": 26, "indvandrere": 9},
                      "Undervisning", "Offentlig", "Nordjylland")
print(skolen)
```

## 11: Når gruppen ikke findes

28 af de mulige kombinationer mangler i tallene, fordi der ville være for få
arbejdssteder i dem.

Så må vi spørge bredere. I tabellerne betyder værdien `Alle`, at vi ser bort fra den
inddeling. Funktionen prøver først det præcise, så uden sektor, og til sidst kun
branchen.

Funktionen giver **to ting** tilbage: rækken og en tekst om, hvilken gruppe det blev.
Det kan en funktion godt — skriv dem med komma imellem efter `return`.

> *Think Python kapitel 6: Return Values.*

**Udfyld hullerne.**

```python
def find_gruppe(tal, dimension_id, branche, stoerrelse, sektor, aar=2022):
    """Find den mest præcise sammenligningsgruppe, der findes i tallene."""

    forsoeg = [
        (stoerrelse, sektor, f"{branche}, {stoerrelse} ansatte, {sektor.lower()} sektor"),
        (stoerrelse, "Alle", f"{branche}, {stoerrelse} ansatte, begge sektorer"),
        ("Alle", "Alle", f"{branche}, alle størrelser"),
    ]

    for stoerrelse_nu, sektor_nu, beskrivelse in forsoeg:
        raekker = tal[(tal["aar"] == aar)
                      & (tal["dimension_id"] == dimension_id)
                      & (tal["branche"] == branche)
                      & (tal["stoerrelse"] == stoerrelse_nu)
                      & (tal["sektor"] == sektor_nu)]

        if len(raekker) _______ 0:
            return raekker.iloc[0], _______

    return None, "ingen gruppe fundet"


raekke, beskrivelse = find_gruppe(tal, 1, "Industri", "50-99", "Privat")
print(beskrivelse)
print("median:", raekke["p50"])
```

```python
# tjek
_, hvad = find_gruppe(tal, 1, "Undervisning", "100-249", "Offentlig")
print("Fandt:", hvad)
```

## 12: Hvor ligger arbejdspladsen?

Nu har vi en andel og en fordeling. Tilbage er at sige, hvor i fordelingen andelen
ligger.

`p25` er det tal, en fjerdedel af arbejdspladserne ligger under. `p75` er det tal, tre
fjerdedele ligger under. Ligger vi under `p25`, er vi altså blandt den laveste
fjerdedel.

Bemærk `elif`: kun den **første** gren, der passer, bliver kørt.

**Udfyld hullerne.**

```python
def placering(andel, raekke):
    """Hvor ligger andelen i fordelingen: lav, middel eller høj?"""
    if andel < raekke[_______]:
        return "i den laveste fjerdedel"
    elif andel < raekke[_______]:
        return "i midten"
    else:
        return "i den højeste fjerdedel"


raekke, beskrivelse = find_gruppe(tal, 1, "Undervisning", "100-249", "Offentlig")

print(f"Jeres kvindeandel er {skolen.andel('kvinder'):.0%}.")
print(f"Det er {placering(skolen.andel('kvinder'), raekke)} blandt: {beskrivelse}.")
```

```python
# tjek
if placering(0.10, {"p25": 0.5, "p75": 0.8}) == "i den laveste fjerdedel" \
   and placering(0.95, {"p25": 0.5, "p75": 0.8}) == "i den højeste fjerdedel":
    print("Rigtigt.")
else:
    print("Ikke endnu. Den laveste fjerdedel ligger under p25.")
```

## 13: Hele barometeret på én gang

Nu sætter vi det hele sammen: én funktion, der tager en arbejdsplads og skriver de tre
linjer ud.

Læg mærke til, at der ikke står nogen udregning i den. Den beder bare de dele, I
allerede har lavet, om at gøre hver deres ting. **Det er hele pointen med at dele
tingene op.**

Der er ingen huller i den her. Kør den, og se jeres barometer virke for første gang.

```python
DIMENSIONER = {1: "kvinder", 2: "seniorer", 3: "indvandrere"}


def vurder(arbejdsplads, tal):
    """Skriv, hvor en arbejdsplads ligger på alle tre dimensioner."""
    print(arbejdsplads)
    print("-" * 60)

    for dimension_id, navn in DIMENSIONER.items():
        egen_andel = arbejdsplads.andel(navn)
        raekke, beskrivelse = find_gruppe(tal, dimension_id, arbejdsplads.branche,
                                          "100-249", arbejdsplads.sektor)
        hvor = placering(egen_andel, raekke)
        print(f"{navn:<13} {egen_andel:>5.0%}   {hvor} blandt {beskrivelse}")


vurder(skolen, tal)
```
